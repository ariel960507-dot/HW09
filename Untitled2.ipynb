```python
# ==========================================
# 輸入資料（測試一：需要測試二、三時，直接替換以下兩行字串即可）
# ==========================================
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# ==========================================
# 1. 讀取活動名稱，建立兩份整理後的串列
# ==========================================
# 使用 tuple 拆包取得兩個活動名稱
board_activity, movie_activity = activity_names

# 先移除半形空白、統一轉小寫，再用逗號拆成 list
# 注意：此時保留原本順序與重複項目，且不覆蓋原始的 board_raw 與 movie_raw
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# ==========================================
# 2. 使用集合去重並比較名單
# ==========================================
# 將串列轉成 set，自動去除同一場活動內的重複報名
board_set = set(board_list)
movie_set = set(movie_list)

# 集合運算：
both_set = board_set & movie_set          # 交集 (&)：兩場都有參加
all_set = board_set | movie_set           # 聯集 (|)：至少參加一場（全部不重複名單）
only_board_set = board_set - movie_set    # 差集 (-)：只參加桌遊，沒參加電影
only_movie_set = movie_set - board_set    # 差集 (-)：只參加電影，沒參加桌遊
# ==========================================
# 3. 建立排序串列，使用切片取得前兩位
# ==========================================
# 各自去重並依字母排序
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)

# 全部參加者排序，並使用切片取得前兩位作為「通知名單預覽」
all_sorted = sorted(all_set)
notify_preview = all_sorted[:2]

# ==========================================
# 4. 使用字典保存摘要，計算人數與應收費用
# ==========================================
board_count = len(board_set)
movie_count = len(movie_set)
board_fee = unit_fees[board_activity]
movie_fee = unit_fees[movie_activity]

board_revenue = board_count * board_fee
movie_revenue = movie_count * movie_fee

# 建立名為 summary 的字典保存數字結果
summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": board_count + movie_count,
    "桌遊應收": board_revenue,
    "電影應收": movie_revenue,
    "總應收": board_revenue + movie_revenue,
}

# ==========================================
# 5. 輸出完整摘要
# ==========================================
print("=== 活動資訊與每人費用 ===")
print(f"{board_activity}：每人 {unit_fees[board_activity]} 元")
print(f"{movie_activity}：每人 {unit_fees[movie_activity]} 元\n")

print("=== 1. 整理後串列（保留順序與重複項目） ===")
print(f"{board_activity}整理串列：{board_list}")
print(f"{movie_activity}整理串列：{movie_list}\n")

print("=== 2. 各場去重排序名單與人數 ===")
print(f"{board_activity}名單（共 {summary['桌遊人數']} 人）：{board_sorted}")
print(f"{movie_activity}名單（共 {summary['電影人數']} 人）：{movie_sorted}\n")

print("=== 3. 參加情況比較 ===")
print(f"共同參加者（共 {summary['共同人數']} 人）：{both_sorted}")
print(f"只參加{board_activity}者：{only_board_sorted}")
print(f"只參加{movie_activity}者：{only_movie_sorted}\n")

print("=== 4. 總體統計與通知預覽 ===")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"總報名人次：{summary['報名人次']} 人次")
print(f"通知名單預覽（前兩位）：{notify_preview}\n")

print("=== 5. 應收費用摘要 ===")
print(f"{board_activity}應收：{summary['桌遊應收']:.2f} 元")
print(f"{movie_activity}應收：{summary['電影應收']:.2f} 元")
print(f"總應收金額：{summary['總應收']:.2f} 元")
```
